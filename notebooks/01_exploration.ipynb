```python
print(df["FTHG"].mean())
print(df["FTAG"].mean())
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.histplot(data = df, x ="FTHG", color = "blue", label = "Gospodarze (Home)")
sns.histplot(data = df, x = "FTAG", color = "orange", label = "Goscie (Away")
plt.legend()
plt.xlabel("Gole (goals)")
plt.ylabel("Liczba meczy (matches)")

```
